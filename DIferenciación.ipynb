```python
# Vamos a construir el método de diferenciación por diferencias centrales:
import math as mt
import numpy as np
import matplotlib.pyplot as plt
def f(x):
    return mt.sin(x)
def dx_central(f,x,h):
    return (f(x + h/2) - f(x - h/2))/h
dom = np.linspace(-2*mt.pi, 2*mt.pi,50)
ctr = np.zeros(len(dom))
ctr2 = np.zeros(len(dom))
for i in range(0,len(dom)):
    ctr[i] = dx_central(f,dom[i],1)
    ctr2[i] = f(dom[i])
plt.plot(dom,ctr, ".")
plt.plot(dom, ctr2,".")
plt.plot(dom,ctr, alpha = 0.5)
plt.plot(dom, ctr2, alpha =0.5)
plt.grid()
plt.axhline(y = 0.0, xmin= 0.0, color = "black")
plt.axvline(x = 0.0, ymin = 0.0, color = "black")
# aquí, f es la función que en este caso se escogió para ser sin(x), pero puede ser cualquier otra.
# x representa el punto de nuestro interés donde queremos saber cuánto vale la derivada de nuestra función.
# a y b son dos puntos al rededor de nuestro punto de interés x, siendo a un punto anterior y b un punto posterior.
# el programa nos regresa la derivada de f, dada por f', evaluada en nuestro punto de interés. Nos regresa f'(x).

    
```
