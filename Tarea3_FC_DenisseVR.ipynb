```python
# Ejercicio 1: En este ejercicio, importaremos un archivo de datos(manchassolares.txt) los cuáles usaremos para hacer una gráfica
# de número de manchas contra tiempo.
# a) Se leen los datos del archivo divididos en columnas y cada una se guarda por separado en una lista diferente para posteriormente
# graficar manchas en función del tiempo (meses).
import numpy as np
import matplotlib.pyplot as plt
datos = open("manchasolares.txt","r")
meses = []
manchas = []
for linea in datos:
    numeros = linea.split()
    meses.append(numeros[0])
    manchas.append(numeros[1])
Meses = list(map(int,meses))
Manchas = list(map(float,manchas))
# print(meses)
# print(manchas) #<---- estas dos líneas son sólo para corroborar que el archivo se leyó y agrupó las columnas de la manera deseada.
plt.plot(meses,manchas, "g.") 
plt.plot(meses,manchas, alpha =0.5)
plt.xlabel("Meses")
plt.ylabel("Manchas solares")
plt.grid() 
```
