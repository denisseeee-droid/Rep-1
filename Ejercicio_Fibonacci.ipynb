```python
# Fibonacci con recursividad
# fb_n = fb_(n-1) + fb_(n-2)
def fb(n):
    if n == 2 or n==1 :
        return(1)
    else:
        return fb(n-1)+fb(n-2)
for i in range(1,10):
    print(fb(i))
```

```python
# función caótica 
# la ecuación del mapeo logístico se define por : x_(n+1) = rx_n*(1-x_n)
import matplotlib.pyplot as plt
def maplog(n, r, x_0):
    if n == 0:
        return(x_0) # regresa la condición inicial
    else:
        return r*maplog(n-1,r,x_0)*(1-maplog(n-1,r,x_0))
it = []
valores = []
for i in range(0,28):
    it.append(i)
    valores.append(maplog(i,3,1/2))
Iteración = list(map(int,it))
Valor = list(map(float,valores))
plt.plot(Iteración,Valor, "g.")
plt.plot(Iteración,Valor, alpha =0.5)

    
# el índice n nos indica el término de la serie que estamos calculando
# el valor x_0 es la condición inicial del problema
```

```python
# función de Mandelbrot
# similar al ejercicio anterior, pero ahora con numeros completos, está la fórmula de Mandelbrot
# z_(n+1) = z_n² + c
def Mandelbrot(n,z_0,c):
    if n == 0:
        return(z_0)
    else:
        return Mandelbrot(n-1,z_0*z_0,c)**2 + c
for i in range(1,10):
    print(Mandelbrot(i,0,3))
# z_0 = 0 ---> z_1 = 0² + 3 = 3
# z_2 = 9 + 3 = 12
# z_3 = 144 + 3 = 147
# el índice n indica el término de la serie que se obtiene
# el valor z_0 indica la condición inicial
# el valor c indica el valor de la constante compleja
```
